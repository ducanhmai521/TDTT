```python
# Bài 3
a,b,c=map(float,input().split())
if a<(b+c) and b<(a+c) and c<(a+b):
    hp=(a+b+c)/2
    s=(hp*(hp-a)*(hp-b)*(hp-c))**0.5
    if a==b and b==c: print("Tam giac deu")
    elif a==b or a==c or b==c: print("Tam giac can")
    elif a*a+b*b==c*c or a*a+c*c==b*b or b*b+c*c==a*a: print("Tam giac vuong")
    else: print("Tam giac thuong")
    print(f"Chu vi: {a+b+c}")
    print(f"Dien tich: {s:.2f}")
else: print("Khong la ba canh cua mot tam giac")
```

```python
# Bài 5:
import math
a,b,c=map(float,input().split())
delta=b*b-4*a*c
if delta<0: print("PT Vo nghiem")
elif delta==0: print(f"x1=x2={-b/2*a}")
elif delta>0:
    print(f"x1= {((-b+math.sqrt(delta))/2*a):.2f}")
    print(f"x2= {((-b-math.sqrt(delta))/2*a):.2f}")
```
