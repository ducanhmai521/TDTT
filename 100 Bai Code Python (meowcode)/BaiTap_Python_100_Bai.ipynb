### 0006 - Tính chu vi và diện tích của một tam giác
**Yêu cầu:** Viết chương trình tính chu vi (cv) và diện tích (s) một tam giác có 3 cạnh a, b, c là các số nguyên theo công thức cv=a+b+c, s=sqrt(p*(p-a)*(p-b)*(p-c)) với p=cv/2 (3 số a, b, c bảo đảm là 3 cạnh của 1 tam giác).


```python
# Bài 0006: Tính chu vi và diện tích của một tam giác
import math
a,b,c=map(int,input().split())
cv=a+b+c
p=cv/2
s=math.sqrt(p*(p-a)*(p-b)*(p-c)) 
print(f"{cv:.1f} {s:.3f}")
```

### 0033 - Đếm số nguyên tố từ m đến n
**Yêu cầu:** Viết chương trình nhập vào 2 số nguyên m và n (m<n). Đếm xem trong các số nguyên từ m đến n, có bao nhiêu số nguyên tố


```python
# Bài 0033: Đếm số nguyên tố từ m đến n
def snt(n):
    count=0
    for i in range(1,n+1):
        if n%i==0: count+=1
    if count==2: return(True)
    else: return(False)
m,n=map(int,input().split())
res=0
for i in range(m,n+1):
    if snt(i): res+=1
print(res)
```

### 0034 - In các số nguyên tố từ m đến n

```python
# Bài 0034: In các số nguyên tố từ m đến n
def snt(n):
    count=0
    for i in range(1,n+1):
        if n%i==0: count+=1
    if count==2: return(True)
    else: return(False)
m,n=map(int,input().split())
res=[]
for i in range(m,n+1):
    if snt(i): res.append(i)
if res: print(*res, sep=" ")
else: print("-")
```

### 0035 - Tính trung bình cộng các số nguyên tố từ m đến n
**Yêu cầu:** Viết chương trình nhập vào 2 số nguyên m và n (m<n). Tính trung bình cộng các số nguyên tố từ m đến n. Nếu không có số nào thì in dấu -


```python
# Bài 0035: Tính trung bình cộng các số nguyên tố từ m đến n
def snt(n):
    count=0
    for i in range(1,n+1):
        if n%i==0: count+=1
    if count==2: return(True)
    else: return(False)
m,n=map(int,input().split())
res=[]
for i in range(m,n+1):
    if snt(i): res.append(i)
if res:
    tong=0
    for i in res: tong+=i
    print(f"{tong/len(res):.2f}")
else: print("-")
```

### 0039 - Kiểm tra số chính phương
