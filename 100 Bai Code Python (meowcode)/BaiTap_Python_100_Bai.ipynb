### 0006 - Tính chu vi và diện tích của một tam giác
**Yêu cầu:** Tính chu vi và diện tích của một tam giác. Hãy viết chương trình xử lý đúng các yêu cầu và định dạng dữ liệu vào/ra.


```python
# Bài 0006: Tính chu vi và diện tích của một tam giác
import math
a,b,c=map(int,input().split())
cv=a+b+c
p=cv/2
s=math.sqrt(p*(p-a)*(p-b)*(p-c)) 
print(f"{cv:.1f} {s:.3f}")
```

### 0008 - Tính bán kính R ngoại tiếp tam giác
**Yêu cầu:** Tính bán kính R ngoại tiếp tam giác. Hãy viết chương trình xử lý đúng các yêu cầu và định dạng dữ liệu vào/ra.


```python
# Bài 0008: Tính bán kính R ngoại tiếp tam giác
import math
a,b,c=map(int,input().split())
cv=a+b+c
p=cv/2 
s=math.sqrt(p*(p-a)*(p-b)*(p-c))
res=(a*b*c)/(4*s)
print(f"{res:.3f}")
```

### 0013 - Tìm số lớn nhất trong 3 số
**Yêu cầu:** Tìm số lớn nhất trong 3 số. Hãy viết chương trình xử lý đúng các yêu cầu và định dạng dữ liệu vào/ra.


```python
# Bài 0013: Tìm số lớn nhất trong 3 số
a,b,c=map(int,input().split())
if a==b==c: print("=")
else:
    toNhat=a
    for i in [a,b,c]:
        if toNhat < i: toNhat=i
    print(toNhat)
```
