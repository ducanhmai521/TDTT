### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def prob4(n):
    count = 0
    for k in range(1, n):
        if k > (n - 1) / 2:
            count = count + 1
    return count
print(prob4(9))
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
def prob6(a,b,c):
    if a<(b+c) and b<(a+c) and c<(a+b):
        p=(a+b+c)/2
        ans=(p*(p-a)*(p-b)*(p-c))**0.5
        return ans
    else: raise ValueError("Sides cannot form a triangle")
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def prob8(letter,n):
    if letter.isupper()==True:
        return chr((ord(letter)-ord("A")+n)%26+ord("A"))
    else: return chr((ord(letter)-ord("a")+n)%26+ord("a"))
print(prob8('a',2))
print(prob8('y',3))
print(prob8('Z',1))
print(prob8('X',5))
print(prob8('A',26))
print(prob8('m',0))
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
def prob12(n):
    return "\n".join(["*"*x for x in range(1,n)])
print(prob12(100))
```
