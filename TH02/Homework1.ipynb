```python
# Bài 8
name=input("tên: ")
thangtruoc=int(input("Thang truoc: "))
thangnay=int(input("Thang nay: "))
print("Ho va ten: "+name)
kw=thangnay-thangtruoc
if kw<=50: moni=(1984*kw)*1.08
elif kw>=51 and kw<=100: moni=((50*1984)+(2050*(kw-50)))*1.08
elif kw>=101 and kw<=200: moni=((50*1984)+(50*2050)+(2380*(kw-100)))*1.08
elif kw>=201 and kw<=300: moni=((50*1984)+(50*2050)+(100*2380)+(2998*(kw-200)))*1.08
elif kw>=301 and kw<=400: moni=((50*1984)+(50*2050)+(100*2380)+(2998*100)+(3350*(kw-300)))*1.08
elif kw>=401: moni=((50*1984)+(50*2050)+(100*2380)+(2998*100)+(3350*100)+(3460*(kw-400)))*1.08
print("Tien phai tra la: ", round(moni))
```
